```python
from meteostat import Stations
from datetime import datetime
```

```python
# Italy
stations = Stations()
stations = stations.nearby(41.8967, 12.4822)
stations = stations.inventory('hourly', datetime(2016, 1, 1))
station = stations.fetch(5)
print(station)
```
